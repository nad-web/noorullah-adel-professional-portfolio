```python
import requests
import json

# ROR-ID der Justus-Liebig-Universität Gießen
ror_id = "https://ror.org/033eqas34"

# Zeitraum
start_year = 2019
end_year = 2022

# API-Endpunkt
url = "https://api.openalex.org/works"

# Abfragen
abfragen = [
    {"filter": f"institutions.ror:{ror_id},publication_year:{start_year}-{end_year}"},
    {"filter": f"institutions.ror:{ror_id},publication_year:{start_year}-{end_year},is_oa:true"},
    {"filter": f"institutions.ror:{ror_id},publication_year:{start_year}-{end_year},oa_status:gold"},
    {"filter": f"institutions.ror:{ror_id},publication_year:{start_year}-{end_year},oa_status:hybrid"}
]

# Ergebnisse
ergebnisse = []

# Durchführen der Abfragen
for abfrage in abfragen:
    response = requests.get(url, params=abfrage)
    response.raise_for_status()
    data = response.json()
    ergebnisse.append(data["meta"]["count"])

# Ausgabe
print("Ergebnisse:")
print(f"Anzahl an Publikationen: {ergebnisse[0]}")
print(f"Anzahl an Publikationen im Open Access (OA): {ergebnisse[1]}")
print(f"Anzahl an Publikationen in Gold OA: {ergebnisse[2]}")
print(f"Anzahl an Publikationen in Hybrid OA: {ergebnisse[3]}")

# Berechnung der Anteile
anteile = []
for i in range(1, len(ergebnisse)):
    anteil = (ergebnisse[i] / ergebnisse[0]) * 100
    anteile.append(anteil)

print("\nAnteile:")
print(f"Anteil an Publikationen im Open Access (OA): {anteile[0]:.1f}%")
print(f"Anteil an Publikationen in Gold OA: {anteile[1]:.1f}%")
print(f"Anteil an Publikationen in Hybrid OA: {anteile[2]:.1f}%")
```
