## Kapitel 1 – Grundlagen der Funktionsdefinition

### Erklärung

Funktionen bündeln wiederverwendbare Anweisungen. Definiert wird mit `def name(parameter): ...`. Rückgabewerte werden mit `return` geliefert. Ohne explizites `return` gibt eine Funktion `None` zurück.


### 1.1 - Einfache Funktion mit Rückgabewert


```python
def quadrat(x):
    """Gibt das Quadrat von x zurück."""
    return x * x

ergebnis = quadrat(5)
print(ergebnis)  # 25
```

### 1.2 - Mehrere `return`-Pfade


```python
def vorzeichenwert(x):
    if x > 0:
        return "positiv"
    elif x == 0:
        return "null"
    else:
        return "negativ"

print(vorzeichenwert(-3))
```

## Kapitel 2 – Parameter: Positionen, Schlüsselwörter, Standardwerte

### Erklärung

Parameter können als Positions- oder Schlüsselwortargumente übergeben werden. Standardwerte vereinfachen häufige Aufrufe.


### 2.1 - Standardparameter und Schlüsselwortargumente


```python
def skalieren(x, faktor=1.0, offset=0.0):
    return x * faktor + offset

print(skalieren(10))                    # Standardwerte
print(skalieren(10, faktor=0.5))        # Schlüsselwortargument
print(skalieren(10, faktor=0.5, offset=2))
```

#### Aufgabe 2.1 – Lineare Abbildung

Eine Funktion `linear(x, m=1.0, b=0.0)` definieren, die `m*x + b` berechnet. Zwei unterschiedliche Aufrufe mit Positions- und mit Schlüsselwortargumenten demonstrieren.


### 2.2 - Mehrere Rückgabewerte (Tupel)


```python
def min_max(values):
    """Gibt (minimum, maximum) zurück."""
    if not values:
        return None, None
    mn = values[0]
    mx = values[0]
    for v in values[1:]:
        if v < mn:
            mn = v
        if v > mx:
            mx = v
    return mn, mx

lo, hi = min_max([3, 1, 5])
print(lo, hi)  # 1 5
```

## Kapitel 3 – Seiteneffekte vermeiden

### Erklärung

Funktionen mit Seiteneffekten ändern externe Objekte (z. B. Listen). In vielen Situationen ist ein „reiner“ Ansatz vorteilhaft: Es wird ein neues Ergebnis erzeugt, statt Eingaben zu verändern.


### 3.1 - Listenmutation vs. neue Liste


```python
def anhängen_in_place(xs, wert):
    xs.append(wert)           # verändert xs
    return xs

def angehängt_kopie(xs, wert):
    neu = list(xs)            # Kopie
    neu.append(wert)
    return neu

daten = [1,2]
r1 = anhängen_in_place(daten, 3)
print("original:", daten, "| neu:", r1)

daten2 = [1,2]
r2 = angehängt_kopie(daten2, 3)
print("original:", daten2, "| neu:", r2)
```
