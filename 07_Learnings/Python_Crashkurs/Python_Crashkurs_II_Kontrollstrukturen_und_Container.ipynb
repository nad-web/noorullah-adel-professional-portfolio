## Kapitel 1 – Kontrollstrukturen: Überblick

### Erklärung

Kontrollstrukturen steuern den Programmfluss. In diesem Abschnitt werden Entscheidungsstrukturen (`if`) sowie Zählschleifen (`for`) und kopfgesteuerte Schleifen (`while`) wiederholt.


### 1.1 if – bedingte Ausführung

Eine Bedingung entscheidet, welcher Block ausgeführt wird.


```python
x = 7
if x > 0:
    print("positiv")
elif x == 0:
    print("null")
else:
    print("negativ")
```

#### Aufgabe 1.1 – Einfache Klassifikation

Eine ganze Zahl in die Klassen „gerade“ oder „ungerade“ einteilen und eine passende Ausgabe erzeugen.


### 1.2 while – wiederholen solange Bedingung erfüllt

Eine `while`-Schleife läuft, solange die Bedingung `True` ist.


```python
count = 0
while count < 3:
    print("Schritt", count)
    count += 1
```

#### Aufgabe 1.2 – Zähler

Mit einer `while`-Schleife eine einfache Rückwärtszählung von 5 bis 1 ausgeben.


### 1.3 for – Iteration über Sequenzen

Mit `for` wird über Elemente einer Sequenz iteriert; `range` erzeugt Zahlenfolgen.


```python
werte = [3, 5, 2]
summe = 0
for w in werte:
    summe += w
print("Summe:", summe)
```

```python
for i in range(3):
    print("Index:", i)
```

#### Aufgabe 1.3 – Aggregation

Aus einer Liste ganzer Zahlen die Summe und das Maximum bestimmen. Ergebnisse ausgeben.

## Kapitel 2 – Listen (`list`)

### Erklärung

Listen sind veränderbare Sequenzen. Häufige Operationen sind Indexierung, Slicing und Methoden wie `append`, `pop`.


### 2.1 Erstellen, Indizieren, Slicing


```python
nums = [10, 20, 30, 40, 50]
print("Erstes Element:", nums[0])
print("Letzte zwei:", nums[-2:])
nums.append(60)
print("Nach append:", nums)
```

#### Aufgabe 2.1 – Listenmanipulation

Eine Liste mit mindestens fünf Zahlen anlegen, das drittletzte Element ausgeben, einen Wert hinten anhängen und anschließend das zweite Element entfernen.


### 2.2 Iteration über Listen


```python
werte = [2, 4, 6, 8]
quadrate = []
for v in werte:
    quadrate.append(v * v)
print("Quadrate:", quadrate)
```

#### Aufgabe 2.2 – Filter

Aus einer Liste ganzer Zahlen nur die geraden Zahlen herausfiltern und in einer neuen Liste speichern.

## Kapitel 3 – Tupel (`tuple`)

### Erklärung

Tupel sind unveränderliche Sequenzen. Typisch sind feste, kleine Datengruppen (z. B. Koordinaten, Parameterpaare) und das **Entpacken** von Werten.


### 3.1 Erstellen und Entpacken


```python
pair = (3, 7)
a, b = pair
print("a:", a, "| b:", b)
```

```python
punkte = [(1, 2), (3, 4), (5, 6)]
for x, y in punkte:
    print(f"Punkt mit x={x}, y={y}")
```

#### Aufgabe 3.1 – Tausch per Tupel

Zwei Variablenwerte per Tupel-Zuweisung vertauschen und das Ergebnis ausgeben.

## Kapitel 4 – Dictionaries (`dict`)

### Erklärung

Dictionaries speichern Zuordnungen von Schlüsseln zu Werten. Übliche Operationen: Zugriff mit `[]`/`get`, Einfügen/Ändern, Iteration über `items()`.


### 4.1 Anlegen, Zugriff, Iteration


```python
noten = {"Alex": 1.7, "Sam": 2.3}
noten["Chris"] = 1.9               # hinzufügen
print("Alex:", noten["Alex"])

for name, note in noten.items():
    print(name, "->", note)
```

#### Aufgabe 4.1 – Auswertung einer Notentabelle

Ein Dictionary von Namen auf Punktzahlen anlegen und den Durchschnitt berechnen. Ergebnis ausgeben.


### 4.2 Sicherer Zugriff mit `get`


```python
wert = noten.get("Pat", None)
print("Eintrag für 'Pat':", wert)  # None, falls Schlüssel nicht vorhanden
```

#### Aufgabe 4.2 – Fehlende Schlüssel behandeln

Bei der Abfrage eines nicht vorhandenen Schlüssels einen Standardwert nutzen und eine passende Meldung ausgeben.

## Kapitel 5 – Sets (`set`)

### Erklärung

Sets enthalten einzigartige Elemente ohne Reihenfolge. Häufige Operationen: Zugehörigkeitstest, Vereinigung (`|`), Schnittmenge (`&`), Differenz (`-`).


### 5.1 Duplikate entfernen und Mengenoperationen


```python
labels = ["A", "B", "A", "C", "B"]
unique_labels = set(labels)
print("Einzigartige Labels:", unique_labels)
```

```python
train = {"A", "B", "C"}
test = {"B", "C", "D"}
print("Schnitt:", train & test)
print("Vereinigung:", train | test)
print("Differenz:", train - test)
```

#### Aufgabe 5.1 – Einzigartige Elemente

Aus einer Liste mit Wiederholungen die Menge der einzigartigen Elemente bestimmen und ausgeben.

## Kapitel 6 – Nützliche Muster

### Erklärung

Es tauchen oft wiederkehrende Muster auf: Iteration über Daten, Sammeln einfacher Statistiken und das Strukturieren kleiner Parameter-Sammlungen.


### 6.1 Schleifenmuster (epochenweise Struktur)


```python
werte = [0.4, 0.35, 0.31]  # beispielhafte „Fehlerwerte“ je Epoche
for epoch, fehler in enumerate(werte):
    print(f"Epoche {epoch}: Fehler={fehler}")
```

#### Aufgabe 6.1 – Laufende Statistik

Über eine Liste von Zahlen iterieren und unterwegs Minimum, Maximum und Durchschnitt (fortlaufend) ausgeben.


### 6.2 Parametrisierung mit Dictionary


```python
params = {"learning_rate": 0.1, "epochs": 3}
for i in range(params["epochs"]):
    # Platzhalter für Verarbeitung
    pass
print("Parameter:", params)
```
