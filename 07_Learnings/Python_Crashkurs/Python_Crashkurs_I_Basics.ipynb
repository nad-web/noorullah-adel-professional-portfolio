### Überblick

Im Fokus stehen grundlegende Datentypen (`int`, `float`, `str`, `bool`) sowie zentrale Operatoren für arithmetische Berechnungen und Vergleiche.


### 2.1 Grundlegende Datentypen

Die Funktion `type(...)` zeigt den Datentyp einer Variable oder eines Wertes. Dadurch lässt sich Verhalten von Ausdrücken besser einordnen.


```python
a = 5          # int
b = 3.2        # float
c = "Hallo"    # str
d = True       # bool

print(type(a), type(b), type(c), type(d))
```

### 2.2 Arithmetische Operatoren

Die wichtigsten Operatoren für numerische Ausdrücke sind `+`, `-`, `*`, `/`, `//` (ganzzahlige Division), `%` (Rest) und `**` (Potenz).


```python
x = 10
y = 55

print("Addition:", x + y)
print("Subtraktion:", x - y)
print("Multiplikation:", x * y)
print("Division:", x / y)
print("Ganzzahldivision:", x // y)
print("Rest:", x % y)
print("Potenz:", x ** y)
```

#### Aufgabe 2.2 – Flächen- und Umfangsberechnung

### 2.3 Vergleiche und boolesche Ausdrücke

Vergleichsoperatoren (`==`, `!=`, `<`, `<=`, `>`, `>=`) liefern Wahrheitswerte. Logische Verknüpfungen erfolgen mit `and`, `or`, `not`.


```python
n = 5
m = 7

print("n == m:", n == m)
print("n != m:", n != m)
print("n < m:", n < m)
print("n <= m:", n <= m)
print("n > m:", n > m)
print("n >= m:", n >= m)

x = True
y = False
print("x and y:", x and y)
print("x or y:", x or y)
print("not x:", not x)
```

## Kapitel 3 – Variablen und einfache Anweisungen

### Überblick

Variablen dienen der Zwischenablage von Werten und Ergebnissen. Neben Zuweisungen werden einfache Ausgaben und Formatierungen wiederholt.


### 3.1 Variablenzuweisung und Ausgabe

Eine einfache Berechnung wird in Variablen abgelegt und anschließend ausgegeben. Formatierte Ausgaben erhöhen die Lesbarkeit.


```python
name = "Alex"
punkte = 18
max_punkte = 24

quote = punkte / max_punkte
print(name, "hat", round(quote*100, 1), "% erreicht.")
```

#### Aufgabe 3.1 – Durchschnitt und Prozentsatz

Drei Punktzahlen und die jeweilige Maximalpunktzahl festlegen. Den Durchschnitt in Prozent berechnen und mit sinnvoller Beschriftung ausgeben.


### 3.2 Zeichenketten und Formatierung

Mit `f`-Strings lassen sich Variablen direkt in eine Zeichenkette einbauen. Das erleichtert strukturierte Ausgaben.


```python
vorname = "Sam"
alter = 27
print(f"{vorname} ist {alter} Jahre alt.")
```

#### Aufgabe 3.2 – Formatierte Ausgabe

Eine kurze Statusmeldung mit mindestens zwei Variablen per `f`-String ausgeben (z. B. Name und Ergebnis).


### 3.3 Zuweisung vs. Ausdrucksauswertung

Zuweisungen kopieren Werte, keine Referenzen auf einfache Zahlen. Das folgende Beispiel illustriert dieses Verhalten.


```python
x = 5
y = x
x = 10
print("x =", x, "| y =", y)  # y bleibt 5
```
